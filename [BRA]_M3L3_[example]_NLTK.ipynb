## Quebrando o texto em frases



```python
sentences = sent_tokenize(text, language='portuguese')

print(20 * "-")
print(sentences)
print(20 * "-")
print("Total de Frases: ", len(sentences))
```

## Deletar palavras de interrupção

```python
stop_words = set(stopwords.words('portuguese'))

print(10 * "-", " Palavras de Interrupção ", 10 * "-")
print(stop_words)
print(20 * "-")
```

```python
words = word_tokenize(text)

print( 20 * "-")
print(words)
print(20 * "-")
print("Total de palavras no seu texto: ", len(words))
print(20 * "-")

wc = WordCloud().generate(' '.join([i for i in words]))
plt.imshow(wc)
```

```python
words = [word.lower() for word in words if word.isalpha()]
words = [word for word in words if word not in stop_words]

print(10 * "-", " Palavras em letra minúscula sem palavras de interrupção", 10 * "-")
print(words)
print(20 * "-")
print("Total de palavras no seu texto:", len(words))
print(20 * "-")

wc = WordCloud().generate(' '.join([i for i in words]))
plt.imshow(wc)
```

## Lemmatization

```python
import inspect

def getargspec(func):
    """Compatibilidade com inspect.getargspec via getfullargspec"""
    from collections import namedtuple
    FullArgSpec = inspect.getfullargspec(func)
    ArgSpec = namedtuple('ArgSpec', 'args varargs keywords defaults')
    return ArgSpec(FullArgSpec.args, FullArgSpec.varargs, FullArgSpec.varkw, FullArgSpec.defaults)

inspect.getargspec = getargspec
```

```python
morph = pymorphy3.MorphAnalyzer()
words = [morph.parse(word)[0].normal_form for word in words]

print(10 * "-", " Palavras na forma inicial ", 10 * "-")
print(words)
print(20 * "-")
print("Total de palavras no seu texto: ", len(words))
print(20 * "-")

wc = WordCloud().generate(' '.join([i for i in words]))
plt.imshow(wc)
```

## Calcular a frequência de cada frase


```python
freq_dist = FreqDist(words)
sentence_scores = {}

for i, sentence in enumerate(sentences):
    sentence_words = word_tokenize(sentence.lower())
    sentence_score = sum([freq_dist[word] for word in sentence_words if word in freq_dist])

    sentence_scores[i] = sentence_score

sentence_scores
```

## Filtrando frases por frequência




```python
sorted_scores = sorted(sentence_scores.items(), key=lambda x: x[1], reverse=True)

sorted_scores
```

## Sumarização do formulário

```python
selected_sentences = sorted_scores[:1]
selected_sentences = sorted(selected_sentences)

# Sumarização do formulário
summary = ' '.join([sentences[i] for i, _ in selected_sentences])
print(summary)
```
